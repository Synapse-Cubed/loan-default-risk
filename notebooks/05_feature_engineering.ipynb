```python
# Import Libraries 
import pandas as pd
import numpy as np
```

```python
# Load the cleaned datasets
df = pd.read_csv(r"..\data\processed\reconciled_loan_data.csv")
```

```python
# Inspect dataset
print(df.info())
print(df.shape)
print(df.describe)
```

```python
# Check both date columns
print(df[['approveddate', 'creationdate']].head())
print(f"\nData types:\n{df[['approveddate', 'creationdate']].dtypes}")
```

```python
# Converting each column separately
df['approveddate'] = pd.to_datetime(df['approveddate'])
df['creationdate'] = pd.to_datetime(df['creationdate'])

# Verify
print(f"Data types:\n{df[['approveddate', 'creationdate']].dtypes}")
print(f"\nSample:\n{df[['approveddate', 'creationdate']].head()}")
```

```python
# Check current state
print(f"Current dtype: {df['hist_loan_count'].dtype}")
print(f"Has nulls: {df['hist_loan_count'].isnull().sum()}")
print(f"Sample values:\n{df['hist_loan_count'].head(10)}")
```

```python
# Check age column
print(df['age'].head())
print(df['age'].dtype)
```

```python
# Convert to int data type
df['age'] = df['age'].astype('Int64')
print(df['age'].dtype)
print(df.info())
```

```python
# Set display options to show all columns side by side
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)
print(df.head(20).to_string())
```

```python
# drop longitude and latitude columns
df = df.drop(columns=['longitude_gps', 'latitude_gps'])
```
